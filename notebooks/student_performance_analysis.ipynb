##  Exploratory Data Analysis (EDA)

Exploratory Data Analysis (EDA) is the process of examining a dataset to understand its main characteristics, patterns, and distributions.

In this project, EDA will help us analyze student performance in Mathematics and Portuguese.

### Objectives
- Calculate average and median final grades.
- Identify minimum and maximum grades.
- Understand the distribution of students' final grades.
- Explore relationships between study time, absences, and academic performance.
- Visualize the results using charts and graphs.

**Note:** The analysis identifies patterns and relationships in the data. It does not automatically establish cause-and-effect relationships.

```python
# Mathematics final grade statistics
print("Mathematics - Final Grade (G3)")
print("Mean:", mat_clean["G3"].mean())
print("Median:", mat_clean["G3"].median())
print("Minimum:", mat_clean["G3"].min())
print("Maximum:", mat_clean["G3"].max())

# Portuguese final grade statistics
print("\nPortuguese - Final Grade (G3)")
print("Mean:", por_clean["G3"].mean())
print("Median:", por_clean["G3"].median())
print("Minimum:", por_clean["G3"].min())
print("Maximum:", por_clean["G3"].max())
```

###  Final Grade Distribution

A histogram is used to understand how numerical values are distributed across different ranges.

In this project, histograms will help us understand:
- Which final grades are most common.
- How student grades are distributed.
- Whether grades are concentrated in lower, middle, or higher ranges.
- How Mathematics and Portuguese grade distributions differ.

The final grade is represented by the `G3` column.

```python
# Mathematics final grade distribution

plt.figure(figsize=(8, 5))

plt.hist(mat_clean["G3"], bins=10, edgecolor="black")

plt.title("Distribution of Mathematics Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Portuguese final grade distribution

plt.figure(figsize=(8, 5))

plt.hist(por_clean["G3"], bins=10, edgecolor="black")

plt.title("Distribution of Portuguese Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

###  Average Grade Comparison

In this section, we compare students' average final grades in Mathematics and Portuguese.

A bar chart makes it easier to compare the average performance of the two subjects.

**Objectives:**
- Calculate the average final grade for each subject.
- Display both averages in a single chart.
- Identify which subject has the higher average final grade.

The comparison is descriptive and does not explain why students perform differently in each subject.

```python
# Calculate average final grades

math_average = mat_clean["G3"].mean()
portuguese_average = por_clean["G3"].mean()

print("Mathematics Average Grade:", round(math_average, 2))
print("Portuguese Average Grade:", round(portuguese_average, 2))
```

```python
# Compare average grades

subjects = ["Mathematics", "Portuguese"]
averages = [math_average, portuguese_average]

plt.figure(figsize=(8, 5))

plt.bar(subjects, averages, edgecolor="black")

plt.title("Average Final Grade Comparison")
plt.xlabel("Subject")
plt.ylabel("Average Final Grade")
plt.ylim(0, 20)

plt.show()
```

### Study Time vs Student Performance

In this section, we explore the relationship between students' weekly study time and their final grades.

The dataset contains a `studytime` column that represents weekly study time in four categories:

- **1:** Less than 2 hours per week
- **2:** 2 to 5 hours per week
- **3:** 5 to 10 hours per week
- **4:** More than 10 hours per week

We will calculate the average final grade for each study-time category and visualize the results using a bar chart.

This analysis helps us identify patterns between study time and academic performance. However, it does not prove that study time alone causes higher grades.

```python
# Calculate average final grade for each study-time category

studytime_analysis = mat_clean.groupby("studytime")["G3"].mean()

print(studytime_analysis)
```

```python
# Visualize study time vs average final grade

plt.figure(figsize=(8, 5))

plt.bar(
    studytime_analysis.index,
    studytime_analysis.values,
    edgecolor="black"
)

plt.title("Study Time vs Average Mathematics Final Grade")
plt.xlabel("Study Time Category")
plt.ylabel("Average Final Grade (G3)")

plt.xticks([1, 2, 3, 4])
plt.ylim(0, 20)

plt.show()
```

### School Absences vs Student Performance

In this section, we investigate the relationship between school absences and students' final Mathematics grades.

The dataset contains an `absences` column that records the number of school absences.

**Objectives:**
- Identify the number of absences recorded for students.
- Explore how absences relate to final grades.
- Calculate the correlation between absences and final grades.
- Visualize the relationship using a scatter plot.

Correlation can indicate whether two variables tend to move together. However, correlation does not prove causation.

```python
# Calculate correlation between absences and final grades

absence_correlation = mat_clean["absences"].corr(mat_clean["G3"])

print("Correlation between Absences and Final Grades:")
print(round(absence_correlation, 3))
```

```python
# Visualize absences vs final grades

plt.figure(figsize=(8, 5))

plt.scatter(
    mat_clean["absences"],
    mat_clean["G3"],
    alpha=0.6
)

plt.title("School Absences vs Mathematics Final Grades")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade (G3)")

plt.show()
```

### Student Grade Progression

In this section, we analyze student performance across three grading periods.

The dataset contains three important grade columns:

- **G1:** First-period grade.
- **G2:** Second-period grade.
- **G3:** Final grade.

**Objectives:**
- Calculate the average grade for each period.
- Compare average grades across the three periods.
- Visualize changes in average academic performance.

This analysis describes overall grade patterns. It does not necessarily mean that every individual student's grades improved or declined.

```python
# Calculate average grades for each period

g1_average = mat_clean["G1"].mean()
g2_average = mat_clean["G2"].mean()
g3_average = mat_clean["G3"].mean()

print("First Period Average (G1):", round(g1_average, 2))
print("Second Period Average (G2):", round(g2_average, 2))
print("Final Period Average (G3):", round(g3_average, 2))
```

```python
# Compare average grades across three periods

periods = ["G1 - First Period", "G2 - Second Period", "G3 - Final Period"]
average_grades = [g1_average, g2_average, g3_average]

plt.figure(figsize=(9, 5))

plt.bar(periods, average_grades, edgecolor="black")

plt.title("Average Mathematics Grades Across Three Periods")
plt.xlabel("Grading Period")
plt.ylabel("Average Grade")
plt.ylim(0, 20)

plt.show()
```

###  Correlation Heatmap

A correlation heatmap is a visualization that represents the correlation between numerical variables using colors and numerical values.

In this project, we will examine relationships between selected variables, including:

- First-period grade (`G1`)
- Second-period grade (`G2`)
- Final grade (`G3`)
- Study time (`studytime`)
- School absences (`absences`)
- Free-time availability (`freetime`)

**Objectives:**
- Identify variables with stronger positive or negative linear correlations.
- Examine relationships between grades from different periods.
- Explore possible relationships between study time, absences, and final grades.

Correlation does not establish causation. Also, some variables, such as `studytime` and `freetime`, are categorical ratings, so their correlations should be interpreted cautiously.

```python
# Select numerical columns for correlation analysis

selected_columns = [
    "G1",
    "G2",
    "G3",
    "studytime",
    "absences",
    "freetime"
]

correlation_data = mat_clean[selected_columns]

# Calculate correlation matrix
correlation_matrix = correlation_data.corr()

# Display correlation values
print(correlation_matrix.round(2))
```

```python
# Create a correlation heatmap

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap - Student Performance")
plt.show()
```
